```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image 

print("Libraries imported successfully!")
```

```python
paddy_data_path = 'E:/LeafLens_Project/data/paddy-disease-classification/train_images'
```

```python
import os
import pandas as pd

paddy_categories = os.listdir(paddy_data_path)

paddy_categories = [c for c in paddy_categories if os.path.isdir(os.path.join(paddy_data_path, c))]


paddy_counts = {}


for category in paddy_categories:
    category_path = os.path.join(paddy_data_path, category)

    num_images = len(os.listdir(category_path))
    paddy_counts[category] = num_images


paddy_df = pd.DataFrame(list(paddy_counts.items()), columns=['Disease', 'Image Count'])

print("Paddy Dataset Image Counts:")
print(paddy_df)
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='Image Count', y='Disease', data=paddy_df, palette='viridis')
plt.title('Number of Images per Paddy Disease Category')
plt.xlabel('Number of Images')
plt.ylabel('Disease Category')
plt.show()
```

```python
def display_sample_images(data_path, categories, num_samples=3):
    plt.figure(figsize=(15, 5))
    for i, category in enumerate(categories[:num_samples]):
        category_path = os.path.join(data_path, category)
        sample_image_name = os.listdir(category_path)[0]
        sample_image_path = os.path.join(category_path, sample_image_name)
        
        img = Image.open(sample_image_path)
        plt.subplot(1, num_samples, i + 1)
        plt.imshow(img)
        plt.title(category)
        plt.axis('off')
    plt.show()

print("Sample Images from Paddy Dataset:")
display_sample_images(paddy_data_path, paddy_categories)
```

```python
# 1. Set the path to your tea leaf data
# IMPORTANT: Change this to the correct path on your computer!
tea_data_path = 'E:/LeafLens_Project/data/tea sickness dataset'

# 2. Count the images in each quality tier
tea_categories = os.listdir(tea_data_path)
tea_counts = {}
for category in tea_categories:
    category_path = os.path.join(tea_data_path, category)
    num_images = len(os.listdir(category_path))
    tea_counts[category] = num_images

tea_df = pd.DataFrame(list(tea_counts.items()), columns=['Quality Tier', 'Image Count'])
print("\nTea Dataset Image Counts:")
print(tea_df)

# 3. Visualize the tea leaf counts
plt.figure(figsize=(10, 5))
sns.barplot(x='Quality Tier', y='Image Count', data=tea_df, palette='plasma')
plt.title('Number of Images per Tea Leaf Quality Tier')
plt.xlabel('Quality Tier')
plt.ylabel('Number of Images')
plt.show()

# 4. Display sample tea leaf images
print("Sample Images from Tea Leaf Dataset:")
display_sample_images(tea_data_path, tea_categories)
```
